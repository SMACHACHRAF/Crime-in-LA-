```python
import numpy as np
target = 'Status'  # Cible choisie
y = df[target]
x = df.drop(columns=[target, 'DR_NO', "Date Rptd", "DATE OCC", "LOCATION", "Status", "Status Desc",'Crm Cd 2', 'Crm Cd 3', 'Crm Cd 4','Cross Street', 'Weapon Used Cd', 'Weapon Desc'])  # On enlève la cible et la colonne inutile 'DR_NO'
print(f"Shape of X: {x.shape}")
print(f"Shape of y: {y.shape}")
```

```python
# Vérifier que les données sont bien chargées
print("Colonnes des caractéristiques : ", x.columns)
```

```python

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
import pandas as pd
# Conversion des variables catégorielles
X = pd.get_dummies(x, drop_first=True)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Créer un modèle d'arbre de décision avec des paramètres pour éviter l'overfitting
model = DecisionTreeClassifier(
    criterion="gini",  # Critère de séparation
    max_depth=3,  # Limiter la profondeur de l'arbre
    min_samples_split=20,  # Minimum d'échantillons pour diviser un noeud
    min_samples_leaf=10,  # Minimum d'échantillons par feuille
    random_state=42
)

# Entraîner le modèle
model.fit(X_train, y_train)

# Évaluer le modèle
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))

# Visualiser l'arbre de décision
plt.figure(figsize=(12, 8))
plot_tree(model, filled=True, feature_names=X.columns, class_names=model.classes_)
plt.show()
```

```python
from sklearn.metrics import classification_report, confusion_matrix

cm_tree = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(10,7))
sns.heatmap(cm_tree, annot=True, fmt="d", cmap="Blues", xticklabels=np.unique(y), yticklabels=np.unique(y))
plt.title("Decision Tree Confusion Matrix")
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()
```
